#  Telecom Customer Churn Prediction — Data Ingestion & Dataset Understanding
This notebook focuses on loading and understanding the raw Telco Customer Churn dataset before any preprocessing or feature engineering.


## Dataset Information
**Dataset Name:** Telco Customer Churn Dataset

**Target Variable**

`Churn`

- Yes → Customer left the company.
- No → Customer stayed with the company.

```python
#Import Libraries 

import pandas as pd
import numpy as np 

pd.set_option("display.max_columns",None)
pd.set_option("display.max_rows",100)

import warnings
warnings.filterwarnings("ignore")
```

##  Load Raw Dataset

```python
DATA_PATH="../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv"
df=pd.read_csv(DATA_PATH)
```

```python
#first five previews 
df.head()
```

##  Dataset Dimensions

```python
rows,columns=df.shape

print(f"Number of rows:{rows}")
print(f"number of columns: {columns}")

print(f"Total Rows and columns {df.shape}")
```

##  Feature Names

```python
columns_df=pd.DataFrame ({"Column Number":range(1,len(df.columns)+1),"Column Names":df.columns})
columns_df
```

##  Dataset Information

```python
df.info()
```

```python
#checking for blank value in Total Charges column if present replacing them with np.nan

df["TotalCharges"]=df["TotalCharges"].replace(" ",np.nan)
```

```python
##  converting Total Charges column from OBJECT to float
df["TotalCharges"]=df["TotalCharges"].astype(float)
```

```python
df.info()
```

## Numerical vs Categorical Features

```python
categorical_columns=df.select_dtypes(include="object").columns.tolist()

numerical_columns=df.select_dtypes( exclude="object").columns.tolist()

print(f"Categorical Features:{len(categorical_columns)}")
print(categorical_columns)

print("\n" + "="*60)

print(f"Numerical Features:{len(categorical_columns)}")
print(numerical_columns)

```

##  Missing Value Analysis


```python
missing_df=pd.DataFrame({"Missing Values":df.isnull().sum(),"Missing Percentage":(df.isnull().sum()/len(df)*100).round(2)})

missing_df.sort_values(by="Missing Percentage",ascending=False)
```

```python
df=df.dropna(subset=["TotalCharges"])
```

```python
#After dropping Missing value column
missing_df
```

##  Duplicate Record Analysis

```python
duplicated_record=df.duplicated().sum()

print(f"Duplicated Count : {duplicated_record}")
```

##  Unique Values Analysis

```python
unique_df=pd.DataFrame({"Data Type":df.dtypes,"Unique Values":df.nunique()})

unique_df.sort_values(by="Unique Values",ascending= False)
```

##  Dataset Memory Usage

```python
memory_usage=df.memory_usage(deep=True).sum()/(1024**2)
print(f"Memory Usage : {memory_usage:.2f} MB")
```

##  Target Variable Distribution

```python
target_summary= pd.DataFrame({"Count":df["Churn"].value_counts(),"Percentage":(df["Churn"].value_counts(normalize=True)*100).round(2)})
target_summary
```

#  Data Quality Report

```python
summary_df=pd.DataFrame({"Data Type":df.dtypes,"Missing Values":df.isnull().sum(),"Missing Percentage":(df.isnull().sum()/len(df)*100).round(2),
                         "Unique Values":df.nunique()})

summary_df
```
